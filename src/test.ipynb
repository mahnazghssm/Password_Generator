```python
from abc import ABC, abstractmethod

class PasswordGenerator(ABC):
    @abstractmethod
    def generate(self):
        pass
```

```python
import random
import string

class PinGenerator(PasswordGenerator):
    def __init__(self, length):
        self.length = length

    def generate(self):
        return "".join([random.choice(string.digits) for _ in range(10)])
```

```python
import random
import string

class RandomPasswordGenerator(PasswordGenerator):
    def __init__(self, length= 10, include_numbers= False, include_symbols= False):
        self.length = length
        self.characters = string.ascii_letters
        if include_numbers:
            self.characters += string.digits
        if include_symbols:
            self.characters += string.punctuation

    def generate(self):
        return "".join(random.choice(self.characters) for _ in range(self.length))
```

```python
r = RandomPasswordGenerator(length = 20, include_numbers= True, include_symbols=True)
r.generate()
```

```python
import nltk

class MemorablePassword(PasswordGenerator):
    def __init__ (self, num_of_words = 6, separator= "_", capitalize = False):
        self.num_of_words = num_of_words
        self.separator = separator
        self.capitalize = capitalize
        self.vocabulary = nltk.corpus.words.words()
    def generate(self):
        memorable_password = [random.choice(self.vocabulary) for _ in range(self.num_of_words)]
        if self.capitalize:
            memorable_password = [word.upper() if random.choice([True, False]) else word.lower() for word in memorable_password]

        return self.separator.join(memorable_password)
```

```python
m = MemorablePassword()
m.generate()
```
